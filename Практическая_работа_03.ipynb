## Задание

Напишите функцию, которая по трем переданным натуральным
числам возвращает могут ли эти числа являться членами
арифметической прогрессии. Если «да», то функция возвращает
True, в противном случае – False.

## Решение

### Объявление функции

```python
def prg(a,b,c):
  return a-b == b-c
```

### Демонстрация использования функции

```python

x, y, z = 2,5,8
if prg(x,y,z):
  print(f'прогрессия со сторками {x}, {y}, {z} сущетсвует')
else:
  print(f'прогрессии со сторками {x}, {y}, {z} не сущетсвует')

x, y, z = 1,2,4
if prg(x,y,z):
  print(f'прогрессия со сторками {x}, {y}, {z} сущетсвует')
else:
  print(f'прогрессии со сторками {x}, {y}, {z} не сущетсвует')
```

## Ответы на контроьлные вопросы

1. Чем отличается функция с возвратом значения от функции без возврата значения?

С возвратом значения — использует оператор return, который передаёт результат вычислений обратно в точку вызова. Этот результат можно сохранить в переменную или использовать в выражении.
Без возврата значения — выполняет действия (печатает текст, изменяет данные и т.д.), но не возвращает явного результата.



2. Что такое библиотечная функция?

Это функция, которая уже написана разработчиками языка или сторонними авторами и входит в состав библиотек (модулей). Программисту не нужно писать её код самому — достаточно подключить библиотеку и вызвать функцию, например import sys



3. Почему библиотечные функции похожи на "черные ящики"?

Потому что для их использования достаточно знать имя функции, какие аргументы она принимает и что возвращает. Внутренняя реализация скрыта от пользователя — мы не видим, как именно работает функция, а просто получаем результат.



4. Что делает инструкция x = random.randint(1, 100)?

генерирует рандомное число от 1 до 100 и записывает его в переменную х



5. Что делает инструкция print(random.randint(1, 20))?

выводит рандомное число от 1 до 20



6. Для чего используется вызов random.seed(20)?

Он инициализирует генератор случайных чисел так называемым «зерном» (seed). Это обеспечивает воспроизводимость последовательности случайных чисел: при каждом запуске программы с random.seed(20) последовательность будет одинаковой.



7. Каким образом функции помогают повторно использовать код?

Функция позволяет один раз написать блок кода для решения задачи, а затем вызывать его многократно из разных мест программы. Это избавляет от копирования кода, делает программу короче, понятнее и упрощает её модификацию — исправлять нужно только в одном месте.



8. Назовите и опишите две части определения функции.

Заголовок — начинается с ключевого слова def, содержит имя функции и список параметров в скобках, заканчивается двоеточием

Тело функции — блок кода с отступом, который выполняется при вызове. Содержит инструкции и, возможно, оператор return.



9. Как задать параметры по умолчанию?

Параметры по умолчанию задаются в заголовке функции с помощью знака равенства. Если при вызове аргумент не передан, используется значение по умолчанию.
Пример: def greet(name="Гость"): — при вызове greet() имя будет «Гость».




10. что такое рекурсия?
Рекурсия — это когда функция вызывает саму себя бесконечное число раз. чтобы ее остановить нужно условие, при котором она останавливается
