20. Что такое язык ассемблера?

Язык ассемблера — это язык программирования низкого уровня, который является символическим (текстовым) представлением машинного кода. Каждая команда ассемблера соответствует одной конкретной машинной инструкции процессора. Он использует мнемонические коды (например, MOV, ADD, SUB) вместо двоичных чисел, что упрощает написание программ, но при этом код остается жестко привязанным к архитектуре конкретного процессора.
